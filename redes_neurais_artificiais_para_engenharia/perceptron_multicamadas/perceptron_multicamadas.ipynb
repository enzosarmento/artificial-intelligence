### 5.7 - Exercícios

1. Explique se é possível realizar o treinamento da rede PMC, por meio do algoritmo backpropagation, quando se inicializa todas as matrizes de pesos com elementos nulos. Discorra também se há então alguma implicação quando se inicializa todos os elementos das matrizes de pesos com valores iguais (diferentes de zeros).

    R: Não é possível realizar o treinamento da rede PMC quando inicializado os pesos com elementos nulos, todos os neurônios irão produzir a mesma saída independente dos valores da entrada e não aprendem. Ainda que inicialize com valores diferentes de zero e sendo o mesmo valor para todos os pesos, continuará o problema da simetria em que os neurônios serão identicos.

2. Considerando os problemas envolvendo aproximação de funções, discorra então se há alguma vantagem e/ou desvantagem em se utilizar a função de ativação linear para os neurônios da camada de saída da rede ao invés do uso da tangente hiperbólica.

    R: A vantagem da função linear se dá que a camada de saída da rede é ilimitada, podendo produzir qualquer valor dentro do conjunto dos números reais, sendo ideial para problemas de aproximação de funções (regressão), o valor previsto podendo ser qualquer coisa (ex: preço, temperatura, altura). A tangente hiperbólica só pode produzir valores entre -1 e 1, sendo o mais ideal a utilização em problemas de classificação.

3. Explique o que são situações de underfitting e overfitting, descrevendo-se também os meios para as suas detecções e as técnicas utilizadas para o seu contorno.

    R: O underfitting é quando o modelo performa abaixo do esperado, mostrando que o modelo é muito simples para identificar e capturar o padrão presente nos dados. Os erros são altos tanto nos dados de treinamento quanto nos dados de teste.

    O overfitting é de certa forma o oposto, onde o modelo se ajusta excessivamente aos dados; ele passa a decorar e memorizar as respostas exatas, funcionando perfeitamente para os dados de treinamento, mas performando mal para os dados de teste. Existem dois casos principais em que o overfitting acontece: 1. quando o modelo é complexo demais, e 2. quando o modelo treinou por tempo demais.

    No caso 1, a ideia é diminuir essa complexidade, tirando neurônios da camada escondida ou até mesmo excluindo uma camada inteira. Para o caso 2, existe uma técnica chamada "parada antecipada" (early stopping), na qual o treinamento é parado de imediato no momento em que o erro dos dados de treino continua caindo, mas o erro dos dados de teste começa a subir. Assim, o treinamento para no ponto ideal, antes que o modelo comece a decorar em vez de aprender.

4. Discorra sobre as eventuais consequências de se utilizar um valor muito elevado para o termo de aprendizagem {η} em contraposição a um valor muito pequeno para o termo de momentum {a}.

    R: Utilizar um valor muito alto para o termo de aprendizagem fará com que o treinamento dê muitos "saltos" ao redor da solução e nunca convergir de fato. E utilizando um valor muito pequeno para o termo de momentum fará com que a velocidade para o treinamento seja insignificante, anulando o efeito do momentum.

5. Seja um problema de classificação de padrões composto apenas por duas classes, sendo que a respectiva fronteira de separação é representada por uma região convexa compacta (fechada e limitada). Estime então qual seria a
quantidade mínima possível de neurônios que poderia estar associado à primeira camada neural escondida.

    R: Um único neurônio só consegue produzir uma reta, portanto para criar uma região convexa compacta seria necessário criarmos um triângulo, para isso a quantidade mínima possível de neurônios associados a primeira camada escondida é 3.

6. Considere para o exercício anterior a situação de que a fronteira de separabilidade entre as classes é agora representada por duas regiões compactas disjuntas, sendo uma convexa e a outra não convexa. Estime então qual
seria a quantidade mínima possível de neurônios tanto para a primeira como para a segunda camada neural escondida.

    R: A primeira camada será responsável por produzir todas as "linhas" necessárias para produzir ambas as regiões, uma convexa e outra não convexa, a região convexa precisa de no mínimo 3 "linhas" e a região não convexa precisa de no mínimo 4 "linhas", logo a primeira camada será necessário no mínimo 7 neurônios. Já a segunda camada será responsável por produzir as regiões, e nós temos 2 regiões, logo precisamos de no mínimo 2 neurônios na segunda camada.

7. Considere a ação de se aplicar o PMC em um problema envolvendo aproximação de funções. As entradas da rede são a temperatura e a pressão, sendo que a saída fornece a quantidade de calor a ser inserida numa caldeira.
O PMC (após o seu treinamento) deverá ser colocado em operação com o objetivo de auxiliar no controle da caldeira. Explique qual seria o primeiro procedimento a ser verificado frente aos novos valores de pressão e temperatura que estarão chegando às suas entradas.

    R: O primeiro procedimento é verificar se os valores (temperatura e pressão) estão dentro dos valores mínimos e máximos que a rede viu durante o treinamento, se os valores estiverem dentro do intervalo a previsão da rede é confiável, se não estiver a previsão não pode ser confiável, já que a rede estaria extrapolando para uma situação que não conhece.

8. Explique se é possível assumir valores nulos para os limiares {θ} de todos os neurônios da última camada de um PMC, quando de sua aplicação em problemas envolvendo aproximação de funções.

    R: Sim, seria possível treinar a rede sem os limiares da última camada de neurônios, porém não é adequado, já que a rede perderia a flexibilidade e pode resultar em um modelo com erro sistemático.

9. Discorra se é possível aplicar o PMC, em problemas envolvendo aproximação de funções, quando se assume a função de ativação linear para todos os neurônios de sua única camada intermediária.

    R: Não é util assumir a função de ativação linear a sua única camada intermediária, já que fazendo isso toda a sua rede neural passa apenas a ser um modelo linear simples, como um único Perceptron e perde a capacidade de lidar com problemas não lineares. A não-linearidade (de funções de ativação como tanh e sigmoide) nas camadas escondidas que dá às redes neurais o seu verdadeiro poder.

10. Discorra sobre a necessidade do pré-processamento das amostras de treinamento e teste, explicitando-se ainda a sua influência na velocidade de treinamento da rede PMC.

    R: É necessário o pré-processamento das amostras para evitar a saturação dos neurônios. Ao normalizar os dados de entrada, nós garantimos que o resultado da soma ponderada (u) tenha mais chances de cair na "região dinâmica" da função de ativação, onde a curva é ingreme.

    Na região dinâmica a derivada de g'(I) é alta, os ajustes nos pesos serão significativos. Desse modo o treinamento será muito mais rápido e significativo. Sem que haja a normalização, a rede passaria parte do tempo obtendo valores pequenos, tornando o treinamento extremamente lento.
